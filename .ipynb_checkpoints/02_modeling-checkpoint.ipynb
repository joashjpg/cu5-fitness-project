# StepUp - A Step Goal Early-Warning Predictor - Modeling

## 1. Problem Statement

**Problem:** I have a step goal of 10k per day. After a long afternoon, I've gotten halfway through that goal. But when I reach home, all I want to do is chill on my couch or just take a nap, then things get busy and I end up staying halfway through that goal for the rest of the day.

**Objective:** What this project aims to do is to predict, based on my current steps taken and time of day, whether I'll be able to reach that goal or not. It’ll also say how much extra walking would improve my chances of reaching the goal in time.

**Why it matters:** When I get home I tend to skip a walk, and the shortfall only shows up when it’s too late to fix. An early warning turns that into something I can still act on.

**Expected outcome:** A trained classifier, an evaluation, and a working frontend that serves a clear purpose. My classifier should beat a simple baseline.

### 2.2 Remove High-Mismatch Users

Before I train my model on data that's present in two different files, I need to make sure that the data I'm using is consistent across files. So I'll need to determine which users have a high disagreement between their daily and minute-level step counts, and drop them before I train.

I'll take some code from 01_eda_cleaning, with some changes to include users with zero mismatches instead of just showing NaN:

```python
mismatches = daily[daily['StepsDifference'] != 0] # Only the rows with a complete agreement
```

```python
total_days = daily.groupby('Id').size() # The total number of days for each Id
mismatch_counts = mismatches['Id'].value_counts() # The number of rows that disagree, for each Id

# Calculate the percentage of disagreement:
mismatch_rate = round(((mismatch_counts.reindex(total_days.index, fill_value=0) / total_days) * 100), 1).sort_values(ascending=False)
mismatch_rate
```

I will drop users whose minute-level and daily steps disagreed on more than 40% of their days, which is where the data shows a clear separation. There's no clear break below the top 3, and dropping more users costs me data I don't have much of, so 40% should be a reasonable threshold.

```python
drop_list = mismatch_rate[mismatch_rate > 40].index
drop_list
```

```python
keep_daily = ~daily['Id'].isin(drop_list)
daily_cleaned = daily[keep_daily].copy()
```

```python
keep_minute = ~minute['Id'].isin(drop_list)
minute_cleaned = minute[keep_minute].copy()
```

## 3. Defining the Prediction Problem

### 3.1 Structure of the Training Table

Here's what the structure of the training table would look like:

| Id | ActivityDate | MinuteOfDay | StepsSoFar | FinalTotal | Reached10k |
|----------|----------|----------|----------|----------|----------|
| A | Apr 20 | 600 | 1,200 | 11,400 | 1 |
| A | Apr 20 | 870 | 3,000 | 11,400 | 1 |
| A | Apr 21 | 1,110 | 700 | 4,300 | 0 |



*Illustrative values (600 = 10:00am, 870 = 2:30pm, 1,110 = 6:30pm)

We'll call this table `checkpoints`. The `checkpoints` table will be built from the minute-level step data. Each row contains a user, a day, and a time of day measured in minutes since midnight. Then we have the amount of steps taken so far, and our target column, which tells us whether that user reached 10k steps or not by the end of the day.

Each day produces one row per checkpoint, every 30 minutes from 8:00am to 9:30pm, so the model sees what “steps so far” looks like at different times of day, and the app needs that because the user can ask at any time.

To prevent leakage: “steps so far” at a time like 3pm must count only minutes *before* 3pm (as far as an actual person could see at that point). If any later minute leaks in, that means the model is peeking at the answer, and it will score very well on the test set but be useless in an actual prediction because it's not actually predicting anything. The checkpoint minute itself shouldn’t be counted (at 3:00pm it'll count up to 2:59pm).

Moreover, `FinalTotal` and `Reached10k` will never be used as features.

### 3.2 Target Variable

The target column is `Reached10k`: 1 if the user’s full-day step total was 10,000 or more, otherwise it's 0. The full-day total is the sum of that day’s minute-level steps (the same source as the features), and only complete days are used (done in 2.3: 814 kept, 28 dropped), so a partial day can’t produce a false 0. I've picked 10k steps as a fixed goal as that's what the project is built around.

`FinalTotal` is stored in the table next to the target.

All of a day’s checkpoints share the same value, since the outcome belongs to the day, not the checkpoint.

## 4. Feature Engineering

```python
minute_cleaned = minute_cleaned.sort_values(by=['Id', 'ActivityMinute'])
```

```python
minute_cleaned['MinuteOfDay'] = minute_cleaned['ActivityMinute'].dt.hour * 60 + minute_cleaned['ActivityMinute'].dt.minute
```

```python
# This gets the total of steps per user-day, but then subtracting it with the steps taken
# in the current minute, which gives us our "steps so far" - steps taken strictly before this minute
minute_cleaned['StepsSoFar'] = minute_cleaned.groupby(['Id', 'ActivityDate'])['Steps'].cumsum() - minute_cleaned['Steps']
```

```python
minute_cleaned['FinalTotal'] = minute_cleaned.groupby(['Id', 'ActivityDate'])['Steps'].transform('sum')
```

```python
minute_cleaned['Reached10k'] = (minute_cleaned['FinalTotal'] >= 10000).astype(int) # Converts boolean to int
```

```python
checkpoints = minute_cleaned[(minute_cleaned['MinuteOfDay'] % 30 == 0) & (minute_cleaned['MinuteOfDay'].between(480, 1290))].copy()
# 480 - 8AM, 1290 - 9:30AM
```

```python
checkpoints = checkpoints[checkpoints['FinalTotal'] > 0] # not keeping rows where no steps were taken the entire day (could be a tracker/sync issue)
```

```python
checkpoints.drop(['ActivityMinute', 'Steps', 'Hour'], axis=1, inplace=True)
```

```python
checkpoints = checkpoints.reset_index(drop=True)
checkpoints
```

```python
checkpoints.isna().sum()
```

## 5. Train/Test Split & Setting a Baseline

### 5.1 Splitting the data

I won't be using sklearn's train_test_split, because it defaults to shuffling rows by random, so the checkpoints of one day could land in both the train and test set. The model would then be tested on days it has already seen during training and the score will be inflated.

As a workaround, I'll choose a cutoff date instead. I'll train on the days before it, and test on the days from it onward. That also matches how the model will actually be used, since it predicts days it hasn't seen.

```python
sorted(checkpoints['ActivityDate'].unique())
```

```python
print(checkpoints['ActivityDate'].nunique())
```

With 30 dates, a test set of 20% is the last 6 days. So I'll make the cut at May 6 to 11 for the test, and the prior will be for train (April 12 to May 5).

```python
is_train = checkpoints['ActivityDate'] < '2016-05-06'
train_df = checkpoints[is_train].copy()
```

```python
is_test = ~is_train # flipped
test_df = checkpoints[is_test].copy()
```

```python
sorted(train_df['ActivityDate'].unique())
```

```python
train_df['ActivityDate'].nunique()
```

```python
sorted(test_df['ActivityDate'].unique())
```

```python
test_df['ActivityDate'].nunique()
```

Now I'll proceed to manually splitting train & test:

```python
X_train = train_df[['MinuteOfDay', 'StepsSoFar']]
X_test = test_df[['MinuteOfDay', 'StepsSoFar']]
y_train = train_df['Reached10k']
y_test = test_df['Reached10k']
```

```python
print(X_train.shape, X_test.shape, y_train.shape, y_test.shape)
print(y_train.mean(), y_test.mean())
```

The split seems good, the shapes line up (17,864 + 3,332 = 21,196)

Test size is about 15.7%, which is a bit under 20% because the last few dates have fewer complete user-days, but is not really a big gap to need a re-split, so I'll just keep it as it is.

Class balance is almost identical with 34.4% in train and 36.1% in test, so the test set represents the same mix of outcomes.

### 5.2 Scoring the baseline

Before training the models, I need baselines for them to beat.

I’ll use two. The first is an always-predict-0 baseline, using sklearn’s DummyClassifier, which shows why accuracy alone is misleading for this problem. Technically its accuracy could also be computed directly as `1 - y_test.mean()`, but the dummy classifier lets me score it with the same metrics as the real models (precision, recall, F1 and confusion matrix), so every comparison uses the same measurements.

The second baseline is a pace-projection rule, which we'll get into once we're done with the DummyClassifier prediction below.

```python
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
```

```python
dummy_clf = DummyClassifier(strategy="most_frequent") # always predict whichever class is most common in the training data, which is 0 here
```

```python
dummy_clf.fit(X_train, y_train) # ignores X_train completely and just remembers the most common label in y_train
```

```python
dummy_pred = dummy_clf.predict(X_test)
```

```python
# Code from Gemini
def score_model(y_true, y_pred, title):
    """
    Prints standard classification metrics and a confusion matrix.

    Parameters:
    y_true (array-like): Ground truth (correct) target values.
    y_pred (array-like): Estimated targets as returned by a classifier.
    title (str): The name or strategy of the classifier for the print header.
    """
    print(f"--- {title} Performance Metrics ---")

    # zero_division=0 just prevents a warning from cluttering the output
    print(f"Accuracy Score : {accuracy_score(y_true, y_pred):.4f}")
    print(f"Precision Score: {precision_score(y_true, y_pred, zero_division=0):.4f}")
    print(f"Recall Score   : {recall_score(y_true, y_pred, zero_division=0):.4f}")
    print(f"F1 Score       : {f1_score(y_true, y_pred, zero_division=0):.4f}")

    print("\n--- Confusion Matrix ---")
    print(confusion_matrix(y_true, y_pred))
```

```python
score_model(y_test, dummy_pred, 'Dummy Classifier')
```

As we can see above, its accuracy is 63.8%, but recall and F1 are 0 because it never predicts that a user will reach 10k, so it would never give anyone a useful warning. This shows that accuracy alone is a poor measure for this problem, and that a model has to do better on precision, recall and F1 as well.

Now we'll do the pace-prediction rule, which is a fair test of whether machine learning adds anything. It asks whether the person is on pace for 10k if they keep the typical shape of a day.

```python
train_df['Progress'] = train_df['StepsSoFar'] / train_df['FinalTotal']
```

```python
train_df.sample(5)
```

```python
progress_curve = train_df.groupby('MinuteOfDay')['Progress'].mean()
```

```python
progress_curve
```

For each training row, I compute Progress `(StepsSoFar / FinalTotal)`, the share of the day’s steps already taken at that checkpoint, and average it for each time of day to get `progress_curve`, the typical shape of a day.

```python
typical_progress = X_test['MinuteOfDay'].map(progress_curve)
typical_progress.isna().sum()
```

```python
projected_total = test_df['StepsSoFar'] / typical_progress
```

```python
projected_total
```

```python
pace_pred = (projected_total >= 10000).astype(int)
```

```python
score_model(y_test, pace_pred, 'Pace Prediction')
```

To predict, I divide a row’s steps so far by the curve value for its time to get a projected daily total, and predict 1 if that reaches 10,000. E.g. if people are typically 50% done by 2pm and someone has 4,000 steps, the projection is 8,000, so the prediction is 0.

The curve is built from training days only, and a test row only supplies its time and steps so far, so there’s no leakage. This is the rule a person would naturally use, so the models need to beat it to show they add value.

Here's what this pace baseline tells me:
- Accuracy is 0.724 against 0.638 for DummyClassifier, so the rule is clearly better, and it actually warns, with recall of 0.52.
- Of the 970 checkpoints it predicted would reach 10k, 64.6% did. Of the 1,204 checkpoints that really reached 10k, it caught 627 (52%) and missed 577.
- The two kinds of mistake are different. 343 were “you can reach it” when the person then missed the goal, and 577 were “you won’t make it” when the person actually would have.

**The bar for the models:** accuracy above 0.724 and F1 above 0.577, and a better confusion matrix.

## 6. Modeling

### 6.1 Selected Models and Why

I've chosen Logistic Regression and Random Forest as the models I'd want to compare.

**Why?**
- **Logistic Regression:** It’s simple and interpretable, and it gives a probability, which fits the “62% likely to reach your goal” output. With these two features it can only draw a straight-line boundary between “will reach” and “won’t”.
- **Random Forest:** It can learn that what counts as “on pace” changes with the time of day, which a straight line can’t do. That’s why the comparison will be interesting (linear vs flexible)

### 6.2 Model 1 - Logistic Regression

Logistic Regression is sensitive to feature scales, so I'll have to scale the features with StandardScaler before training on them.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
lr = LogisticRegression(max_iter=1000, random_state=42)
lr.fit(X_train_scaled, y_train)
```

```python
lr_pred = lr.predict(X_test_scaled)
lr_proba = lr.predict_proba(X_test_scaled)[:, 1] # to get only the probability of reaching 10k from the 2D array
```

```python
score_model(y_test, lr_pred, "Logistic Regression")
```

What I can see from this:
- The model beats the pace baseline on accuracy and precision, and ties on F1 score, so the overall gain is modest.
- It’s more careful about the costly mistake: it says “you can reach it” wrongly 207 times instead of 343. It catches 90.3% of real shortfalls (1,921 of 2,128) against 83.9% for the pace rule.
- It pays for that with slightly lower recall. It misses more days that would have reached 10k, but the share of its “won’t make it” warnings that are unnecessary is almost the same as the baseline’s.
- A modest gain makes sense, I have only two features, and the pace rule is already a strong baseline.

### 6.3 Model 2 - Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf = RandomForestClassifier(max_depth=8, min_samples_leaf=30, n_estimators=200, random_state=42)
rf.fit(X_train, y_train)
```

```python
rf_pred = rf.predict(X_test)
rf_proba = rf.predict_proba(X_test)[:, 1]
```

```python
score_model(y_test, rf_pred, "Random Forest")
```

Comparison table between the pace rule baseline, and the models Logistic Regression, and Random Forest.

| | Pace rule | Logistic Regression | Random Forest |
|---|---|---|---|
| Accuracy | 0.724 | 0.751 | **0.768** |
| Precision | 0.646 | 0.737 | **0.893** |
| Recall | 0.521 | **0.483** | 0.408 |
| F1 | 0.577 | **0.583** | 0.560 |
| Wrong "you will make it" | 343 | 207 | **59** |
| Unnecessary warnings | 577 | 623 | 713 |
| Real shortfalls caught | 83.9% | 90.3% | **97.2%** |

- Random Forest is the most "safe". When it says you’ll reach 10k, it’s right 89% of the time, and it catches 97% of real shortfalls (2,069 of 2,128).
- The price is that it misses more days that would have reached 10k (713). Even so, the share of its unnecessary warnings (about 25.6%) is close to Logistic Regression’s (about 24.5%).
- F1 is lower than Logistic Regression’s because F1 rewards recall. Which model is better depends on which mistake matters more, and for a warning app that’s the wrong “you will make it”, where Random Forest is clearly best.

Random Forest seems like the choice to make here, but both models are judged at the same 0.5 cutoff, which hides how well they rank days. So before I make a final decision, I'll do a quick ROC AUC check on both the models so I can see if Random Forest is actually better or just more cautious at 0.5.

```python
from sklearn.metrics import roc_auc_score
```

```python
roc_auc_score(y_test, lr_proba) # Logistic Regression
```

```python
roc_auc_score(y_test, rf_proba) # Random Forest
```

Random Forest scores 0.775 against 0.743 for Logistic Regression, so it ranks days better and isn’t just more cautious at the 0.5 cutoff.

AUC of 0.5 is a coin flip and 1.0 is perfect, so 0.775 is decent but not great, which is reasonable because it only has two features to train on.

For the app, I’ll use Random Forest. Since the app exists to warn people early, the costly mistake is saying “you will make it” to someone who then misses the goal. Random Forest made that mistake only 59 times on the test set (Logistic Regression: 207, pace rule: 343) and caught 97.2% of real shortfalls.

The trade-off is a lower recall and F1 than Logistic Regression. It gives more unnecessary warnings (713), which only cost the user a short walk. Logistic Regression would give smoother probabilities, but it catches fewer shortfalls.

## 7. Evaluation & Interpretation

### 7.1 Time-of-day Breakdown

```python
# Code from Claude
res = test_df[['MinuteOfDay', 'Reached10k']].copy()
res['Dummy'], res['Pace'], res['LR'], res['RF'] = dummy_pred, pace_pred, lr_pred, rf_pred
res['TimeOfDay'] = pd.cut(res['MinuteOfDay'], bins=[479, 719, 1019, 1290],
                          labels=['Morning', 'Afternoon', 'Evening']) # 8am-11:59, 12-4:59pm, 5pm-9:30pm

models = ['Dummy', 'Pace', 'LR', 'RF']

# a prediction is "correct" when it equals the true label, the mean of True/False is the accuracy
correct = res[models].eq(res['Reached10k'], axis=0)
print(correct.groupby(res['TimeOfDay'], observed=True).mean().round(3))
print(res['TimeOfDay'].value_counts()) # rows per block
```

I divided the test set into three time blocks (Morning: 952 rows, Afternoon: 1,190 rows, Evening: 1,190 rows).

It seems that the model accuracy converges as the day goes on. In the morning, none of the models do much better than the dummy (LR 0.651 and RF 0.673 vs 0.639), so early-morning predictions are close to guessing. In fact, the simple pace rule actually performs worse than the dummy baseline in the morning (0.589 vs 0.639).

**Likely why:** Early-day steps are small and noisy, while late-day steps account for nearly the entire daily total.

**App takeaway:** The afternoon is the most practical window for predictions before the outcome is locked in by evening.

A caveat to consider is that this is based on a relatively small test set evaluated on accuracy alone.

### 7.2 Plots

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay, RocCurveDisplay

ConfusionMatrixDisplay.from_predictions(y_test, rf_pred)
plt.title('Random Forest: confusion matrix (test set)')
plt.show()

fig, ax = plt.subplots()
RocCurveDisplay.from_predictions(y_test, lr_proba, name='Logistic Regression', ax=ax)
RocCurveDisplay.from_predictions(y_test, rf_proba, name='Random Forest', ax=ax, plot_chance_level=True)
plt.show()
```

**Confusion matrix (Random Forest)**
- Rows are the actual outcome and columns are the prediction. The top-left (2,069) are shortfalls correctly flagged, and the top-right (59) is the costly mistake of saying “you’ll make it” when the user doesn’t.
- The bottom-left (713) are the unnecessary warnings when the user would have made it anyway. That's what lowers recall.
- Together this shows the model is built to be cautious: 97% of shortfalls are caught, but it only recognises about 41% of the days that really reach 10k (491 of 1,204). For a warning app that’s an acceptable trade, since the cost is just an extra walk.

**ROC curve**
- Both curves are well above the diagonal chance line, so both models carry real signal. RF is higher overall (AUC 0.78 vs 0.74).
- At low false-positive rates the two curves overlap, so RF’s advantage appears only when I accept more false alarms. With a small test set, the gap is modest.
- Because the curve flattens, lowering the cutoff below 0.5 would catch more successful days at the cost of more false alarms.

## 10. Conclusion & Limitations

**Findings:** On the test set, the Random Forest reached 0.768 accuracy, 0.893 precision and 0.775 ROC AUC. It said “you’ll make it” wrongly only 59 times (Logistic Regression: 207, pace rule: 343) and caught 97.2% of real shortfalls. Accuracy differences between the models shrink through the day, and in the morning none of the models do much better than always predicting 0.

**Was the objective met?** Yes, partly. The Random Forest beat the pace-rule baseline on accuracy, precision, false positives and AUC, but not on F1 (0.560 vs 0.577) or recall (0.408 vs 0.521). I chose it anyway because, for a warning app, a wrong “you’re fine” is the costly mistake, while an unnecessary warning only costs a short walk. The demo works end to end: it gives a live prediction advice on how much extra walking would raise the chance. The calorie regression was not built, and I dropped it to focus on the classifier and the demo.

**Limitations:**
- The data is 30 self-selected users from 2016, so it may not represent other people.
- The same users appear in train and test, so scores may be optimistic for a brand-new user.
- The model uses only two features, which caps its accuracy, and early-morning predictions are close to guessing.
- It only knows 8:00am to 9:30pm and a fixed 10k goal, and the advice assumes about 100 steps a minute and doesn’t know the user’s timetable.
- I excluded 3 high-mismatch users and 42 zero-step days, which reduces the data further.

## 11. Future Improvements
